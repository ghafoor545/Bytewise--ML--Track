# Task 1: Understanding CNNs
# 1.1 What are the advantages of convolutional layers over fully connected layers in image processing tasks?

### **Key Points:**
Convolutional layers preserve the spatial structure of images, unlike fully connected layers.
They reduce the number of parameters by using shared weights (kernels), making models more efficient and less prone to overfitting.
They capture local patterns (edges, textures) in images which are essential for tasks like image recognition.
They allow models to be translation-invariant, meaning the learned features are recognized regardless of their position in the image.


# 1.2 How does pooling help in reducing the computational complexity of a CNN?

**Key Points:**

Pooling layers reduce the dimensionality of feature maps, leading to fewer computations in subsequent layers.
They also help in making the model invariant to small translations, rotations, or distortions in the input image.
This reduction in size (and thus complexity) comes at the cost of some information loss, but it generally improves the model’s efficiency and performance.

# 1.3 Compare different types of pooling layers (max pooling, average pooling).

**Max Pooling:**

Advantages: Captures the most prominent feature in the pooled region, making it useful for detecting sharp features or high-contrast patterns.
Disadvantages: Can ignore some relevant information by only focusing on the maximum value.

**Average Pooling:**

Advantages: More sensitive to the presence of features across the entire region, which can be useful for tasks where overall texture or smoothness is important.
Disadvantages: May not highlight the most important features as effectively as max pooling.

```python
import cv2
import numpy as np

# Function to perform image segmentation
def perform_segmentation(images):
    segmented_images = []
    for img in images:
        # Perform segmentation (e.g., thresholding)
        _, segmented_img = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)
        segmented_images.append(segmented_img)
    return segmented_images

# Function to detect edges within segmented regions
def detect_edges(images):
    edge_detected_images = []
    for img in images:
        # Detect edges (e.g., using Canny edge detector)
        edges = cv2.Canny(img, 100, 200)
        edge_detected_images.append(edges)
    return edge_detected_images

# Segment and detect edges in enhanced cat images
segmented_cat_images = perform_segmentation(enhanced_cat_images)
edges_cat_images = detect_edges(segmented_cat_images)

# Segment and detect edges in enhanced dog images
segmented_dog_images = perform_segmentation(enhanced_dog_images)
edges_dog_images = detect_edges(segmented_dog_images)

# Notification of image segmentation and edge detection
print(f'Segmented {len(segmented_cat_images)} cat images and detected edges.')
print(f'Segmented {len(segmented_dog_images)} dog images and detected edges.')
```

```python
import cv2
import numpy as np

def preprocess_image(img, size=(64, 64)):
    gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    blurred_img = cv2.GaussianBlur(gray_img, (5, 5), 0)
    enhanced_img = cv2.equalizeHist(blurred_img)

    # Define the sharpening kernel
    kernel = np.array([[0, -1, 0],
                       [-1, 5, -1],
                       [0, -1, 0]])
    sharpened_img = cv2.filter2D(enhanced_img, -1, kernel)

    # Resize the image
    resized_img = cv2.resize(sharpened_img, size)

    return resized_img

def extract_features(images):
    features = []
    for img in images:
        processed_img = preprocess_image(img)
        features.append(processed_img.flatten())  # Simple feature extraction by flattening the image
    return np.array(features)
```

```python
# Load training images
cats_folder_path = '/content/cats'
dogs_folder_path = '/content/dogs'

def load_images_from_folder(folder):
    images = []
    for filename in os.listdir(folder):
        img = cv2.imread(os.path.join(folder, filename))
        if img is not None:
            images.append(img)
    return images

cat_images = load_images_from_folder(cats_folder_path)
dog_images = load_images_from_folder(dogs_folder_path)

# Create labels
cat_labels = [0] * len(cat_images)
dog_labels = [1] * len(dog_images)

# Combine images and labels
all_images = cat_images + dog_images
all_labels = cat_labels + dog_labels

# Extract features
features = extract_features(all_images)
labels = np.array(all_labels)

# Split into training and testing sets
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(features, labels, test_size=0.2, random_state=42)

# Train the SVM classifier
from sklearn.svm import SVC

svm_classifier = SVC(kernel='linear')
svm_classifier.fit(X_train, y_train)

# Evaluate the model on the test set
y_pred = svm_classifier.predict(X_test)
from sklearn.metrics import accuracy_score, classification_report

accuracy = accuracy_score(y_test, y_pred)
print(f'Accuracy on the test set: {accuracy}')
print(classification_report(y_test, y_pred))
```

```python
# Function to load and preprocess a single image
def load_and_preprocess_image(image_path):
    img = cv2.imread(image_path)
    if img is None:
        raise ValueError(f"Image at path {image_path} could not be loaded.")
    return preprocess_image(img)

# Paths to the new images
cat_image_path = '/content/images.jpeg'
dog_image_path = '/content/pexels-pixabay-220938.jpg'

# Load and preprocess the new images
new_cat_image = load_and_preprocess_image(cat_image_path)
new_dog_image = load_and_preprocess_image(dog_image_path)

# Extract features
new_cat_image_features = new_cat_image.flatten()
new_dog_image_features = new_dog_image.flatten()

# Combine the features into an array
new_features = np.array([new_cat_image_features, new_dog_image_features])

# Predict using the trained SVM classifier
predictions = svm_classifier.predict(new_features)

# Notification of predictions
print(f'Predictions for the new images: {predictions}')
print(f'Image {cat_image_path} is a {"cat" if predictions[0] == 0 else "dog"}')
print(f'Image {dog_image_path} is a {"cat" if predictions[1] == 0 else "dog"}')

# Debugging: print the features
print(f'Extracted features for cat image: {new_cat_image_features}')
print(f'Extracted features for dog image: {new_dog_image_features}')
```
