# **Part 1:**

Recurrent Neural Networks (RNNs) are designed to handle sequential data by maintaining a hidden state that captures information from previous steps.

# How RNNs Work:

**Hidden State:** Each time step updates the hidden state based on the current input and the previous hidden state.
**Sequence Handling:** Information from the previous step is passed along with the current input to capture temporal dependencies.
# Difference from Feedforward Networks:

**Feedforward Networks:** Process inputs independently without considering sequence or context.
**RNNs:** Handle sequences and maintain context over time, making them suitable for tasks like language modeling and time-series prediction.

`Task: Explain the working of RNNs and how information is passed through the network over time.`

Explanation:

**Hidden State:** RNNs have a loop in their architecture that allows the hidden state to pass from one time step to the next.
**Processing:** At each time step, the RNN takes the current input and the hidden state from the previous time step, processes them, and updates the hidden state.
**Sequence Handling: **This process continues across the entire sequence, enabling the network to maintain context and capture temporal dependencies.




# **Stacking RNN Layers:**

`**Advantages:**`

Increased Capacity: Allows the model to learn more complex patterns and hierarchical representations.
Improved Performance: Better at capturing intricate relationships in sequences.

`**Drawbacks:**`

Vanishing/Exploding Gradients: Deeper layers can struggle with gradient issues during training.
Higher Computational Cost: More layers mean more parameters and longer training times.

`When to Use:`

Stacked RNNs: Use when dealing with complex or long sequences that require capturing deeper patterns.

`Bi-directional RNNs:`

Overview: Consist of two RNNs running in parallel—one processes the sequence forward, and the other processes it backward. Their outputs are combined.

`Advantages:`

Context from Both Directions: Captures dependencies from both past and future in the sequence.
Enhanced Accuracy: Often improves performance on tasks where full sequence context is important.

`Drawbacks:`

Increased Complexity: Doubles the number of parameters, leading to higher training times.

`When to Use:`

`Bi-directional RNNs:` Use when it's beneficial to consider context from both the beginning and end of the sequence, such as in text analysis or speech recognition.

# **Types of RNN Models:**

`Simple RNN`:

Structure: Basic RNN with a single hidden state updated each time step.
Difference: Limited by vanishing gradient issues for long sequences.

`LSTM (Long Short-Term Memory):`

Structure: Uses memory cells and gating mechanisms to handle long-term dependencies.
Difference: Better at capturing long-term dependencies and mitigating vanishing gradients.

`GRU (Gated Recurrent Unit):`

Structure: Similar to LSTM but with fewer gates (update and reset gates).
Difference: More efficient than LSTM with a simpler architecture.

`Bidirectional RNN:`

Structure: Two RNNs processing the sequence in both forward and backward directions.
Difference: Captures context from both directions for improved accuracy.

`Deep RNN:`

Structure: Stacks multiple RNN layers.
Difference: Increases capacity to learn complex patterns but can face gradient issues.

# **Task 1: Implementing a Basic RNN Model**
We'll use the IMDB dataset for sentiment analysis as an example.

Step 1: Import Required Libraries

````python
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
from tensorflow.keras.preprocessing import sequence

# Set random seed for reproducibility
np.random.seed(7)
````

# **Step 2: Load and Preprocess the Data**

````python
# Load the dataset
max_features = 5000  # Number of words to consider as features
max_len = 500  # Cut off texts after this number of words (among the max_features most common words)
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=max_features)

# Pad sequences to ensure uniform input length
x_train = sequence.pad_sequences(x_train, maxlen=max_len)
x_test = sequence.pad_sequences(x_test, maxlen=max_len)
````

# **Step 3: Build the Basic RNN Model**

````python
model = Sequential()
model.add(Embedding(max_features, 32, input_length=max_len))
model.add(SimpleRNN(32))
model.add(Dense(1, activation='sigmoid'))

model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
model.summary()
````

# **Step 4: Train the Model**

````python
history = model.fit(x_train, y_train, epochs=5, batch_size=128, validation_split=0.2)
````

# **Step 5: Evaluate the Model**

````python
score = model.evaluate(x_test, y_test, batch_size=128)
print(f"Test loss: {score[0]}")
print(f"Test accuracy: {score[1]}")
````

# **Task 2: Stacking RNN Layers and Bi-directional RNNs**
Part 1: Stacked RNN
We'll modify the basic RNN model to include multiple RNN layers.
Step 1: Modify the Model to Include Stacked RNN Layers


````python
model_stacked = Sequential()
model_stacked.add(Embedding(max_features, 32, input_length=max_len))
model_stacked.add(SimpleRNN(32, return_sequences=True))
model_stacked.add(SimpleRNN(32))
model_stacked.add(Dense(1, activation='sigmoid'))

model_stacked.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
model_stacked.summary()
````

# **Step 2: Train and Evaluate the Stacked RNN Model**

````python
history_stacked = model_stacked.fit(x_train, y_train, epochs=5, batch_size=128, validation_split=0.2)
score_stacked = model_stacked.evaluate(x_test, y_test, batch_size=128)
print(f"Stacked RNN - Test loss: {score_stacked[0]}")
print(f"Stacked RNN - Test accuracy: {score_stacked[1]}")
````

# **Part 2: Bi-directional RNN**
We'll modify the basic RNN model to a bi-directional RNN.

Step 1: Modify the Model to Bi-directional RNN

````python
from tensorflow.keras.layers import Bidirectional

model_bi = Sequential()
model_bi.add(Embedding(max_features, 32, input_length=max_len))
model_bi.add(Bidirectional(SimpleRNN(32)))
model_bi.add(Dense(1, activation='sigmoid'))

model_bi.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
model_bi.summary()
````

# Step 2: Train and Evaluate the Bi-directional RNN Model

````python
history_bi = model_bi.fit(x_train, y_train, epochs=5, batch_size=128, validation_split=0.2)
score_bi = model_bi.evaluate(x_test, y_test, batch_size=128)
print(f"Bi-directional RNN - Test loss: {score_bi[0]}")
print(f"Bi-directional RNN - Test accuracy: {score_bi[1]}")
````

# **Task 3: Exploring Hybrid Architectures**
We'll combine an RNN with a Convolutional Neural Network (CNN).

Step 1: Modify the Model to a Hybrid CNN-RNN Architecture

````python
from tensorflow.keras.layers import Conv1D, MaxPooling1D

model_hybrid = Sequential()
model_hybrid.add(Embedding(max_features, 32, input_length=max_len))
model_hybrid.add(Conv1D(64, 5, activation='relu'))
model_hybrid.add(MaxPooling1D(pool_size=4))
model_hybrid.add(SimpleRNN(32))
model_hybrid.add(Dense(1, activation='sigmoid'))

model_hybrid.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
model_hybrid.summary()
````

# Step 2: Train and Evaluate the Hybrid Model

````python
history_hybrid = model_hybrid.fit(x_train, y_train, epochs=5, batch_size=128, validation_split=0.2)
score_hybrid = model_hybrid.evaluate(x_test, y_test, batch_size=128)
print(f"Hybrid Model (CNN + RNN) - Test loss: {score_hybrid[0]}")
print(f"Hybrid Model (CNN + RNN) - Test accuracy: {score_hybrid[1]}")
````
